## Instruct variant models:

* The Tokenizer models we have used till now are the "Base" model, For chatbot purpose: there are "Instruct" variant type models.
* Usually models which are trained for to be used in chat specific context will have "instruct" at the end of the model name.
* They have been trained to expect prompts with a particular format that includes system, user and assistant prompts.
* "Apply_chat_template" will convert messages list format to correct input prompt for the model.

```python
tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen3-VL-8B-Instruct")
```

```python
messages = [
    {"role":"system","content":"You are a helpful assistant"},
    {"role":"user","content":"Tell me a joke based on global warming!"}
]

prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt = True)
print(prompt)
```

## Trying out Tokonizer Models:

Phi: microsoft/Phi-4-mini-instruct <br>
DeepSeek: deepseek-ai/DeepSeek-V3.1

```python
tokenizer2 = AutoTokenizer.from_pretrained("microsoft/Phi-4-mini-instruct")
prompt = tokenizer2.apply_chat_template(messages, tokenize=False, add_generation_prompt = True)
print(prompt)
```

```python
text1 = "She bought a two-headed coin, but the universe found a third way for her to lose. By the way I used Hugging Face Tokenizer!"
tokens = tokenizer2.encode(text1)
print(tokens)
print(tokenizer2.batch_decode(tokens))
```

```python
tokenizer3 = AutoTokenizer.from_pretrained("deepseek-ai/DeepSeek-V3.1")
prompt = tokenizer3.apply_chat_template(messages, tokenize=False, add_generation_prompt = True)
print(prompt)
```

```python
tokens = tokenizer3.encode(text1)
print(tokens)
print(tokenizer3.batch_decode(tokens))
```

## Final Comparison between each models:

```python
LLAMA = "meta-llama/Meta-Llama-3.1-8B-Instruct"
PHI4 = "microsoft/Phi-4-mini-instruct"
DEEPSEEK = "deepseek-ai/DeepSeek-V3.1"
QWEN_CODER = "Qwen/Qwen2.5-Coder-7B-Instruct"
```

```python
phi4_tokenizer = AutoTokenizer.from_pretrained(PHI4)

text = "I am curiously excited to show Hugging Face Tokenizers in action to my LLM engineers"
print("Llama:")
tokens = tokenizer.encode(text)
print(tokens)
print(tokenizer.batch_decode(tokens))
print("\nPhi 4:")
tokens = phi4_tokenizer.encode(text)
print(tokens)
print(phi4_tokenizer.batch_decode(tokens))
```

```python
print("Llama:")
print(tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True))
print("\nPhi 4:")
print(phi4_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True))
```

```python
deepseek_tokenizer = AutoTokenizer.from_pretrained(DEEPSEEK)

text = "I am curiously excited to show Hugging Face Tokenizers in action to my LLM engineers"
print(tokenizer.encode(text))
print()
print(phi4_tokenizer.encode(text))
print()
print(deepseek_tokenizer.encode(text))
```

```python
print("Llama:")
print(tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True))
print("\nPhi:")
print(phi4_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True))
print("\nDeepSeek:")
print(deepseek_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True))
```

```python
qwen_tokenizer = AutoTokenizer.from_pretrained(QWEN_CODER)
code = """
def hello_world(person):
  print("Hello", person)
"""
tokens = qwen_tokenizer.encode(code)
for token in tokens:
  print(f"{token}={qwen_tokenizer.decode(token)}")
```

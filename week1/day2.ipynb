# Chat Completions API

### Calling OpenAI model:

```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")
```

## Accessing the OpenAI provided end-point

```python
import requests

headers = {"Authorization": f"Bearer {api_key}","Content-Type":"application/json"}

payload = {
    "model" : "gpt-5-nano",
    "messages": [{"role":"user", "content": "Tell me a computer science joke."}]
}

payload
```

```python
response = requests.post("https://api.openai.com/v1/chat/completions",headers = headers, json = payload)

response.json()
```

```python
response.json()["choices"][0]['message']['content']
```

## OpenAI Package

* A wrapper making the exact same thing which we have done in the above code.
* It allows us to work with python code instead of json objects.

```python
from openai import OpenAI
from IPython.display import Markdown, display
openai = OpenAI()

response = openai.chat.completions.create(
    model = "gpt-5-nano",
    messages = [{
        "role":"user","content":"Tell me a scientific fact about space."
    }]
)

display(Markdown(response.choices[0].message.content))
```

## We can use OpenAI library to also access the http endpoint of other models such as gemini.

Gemini url = "hhtps://generativelanguage.googleapis.com/vibeta/openai/"

```python
Gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

google_api_key = os.getenv("GOOGLE_API_KEY")
```

```python
Gemini = OpenAI(base_url = Gemini_url, api_key = google_api_key)

response = Gemini.chat.completions.create(
    model ="gemini-2.5-flash",
    messages = [{
        "role":"user","content":"Tell me a scientific fact about space."
    }]
)

display(Markdown(response.choices[0].message.content))
```

# Home-Work:

Replicate the day1 homework but with ollama model

```python
from scraping import fetch_website_contents
```

```python
def create_message(website):
    return [
        {"role":"system", "content":"You are a professional game reviewer. Explain the story or context of the game based on the given data."},
        {"role":"user", "content":"Based on the given scraped data about popular game, provide the analysis of the game lore" + website}
            ]
```

```python
requests.get("http://localhost:11434").content

def web_data(url):
    website = fetch_website_contents(url)
    messages = create_message(website)
    
    ollama = OpenAI(base_url = "http://localhost:11434/v1", api_key = "ollama")
    response = ollama.chat.completions.create(
        model = "qwen3:4b",
        messages = messages
    )
    return display(Markdown(response.choices[0].message.content))
```

```python
web_data("https://en.wikipedia.org/wiki/Elden_Ring")
```

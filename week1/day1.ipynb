### Types of Prompts:

* You can pass prompts to the model to shape the way and tone the model generates reponse.
* A system prompt: tells them what task they are performing and what tone they should use.
* A user prompt: The input from the user that the model needs to respond to.

````python
system_prompt = """ you are an assistant that analyzes the content of a website,
and provides a short summary, ignoring text that might be navigation related or not important.
Respond in Markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
user_prompt = """Here are the contents of a news website.
Provide a short summary of this websites markdown. Explain the top headline for todays news.
if it includes news or announcments, then summarize those too.

"""
````

## Messages Format:

* OpenAI's models accept messages in a particular format: <br>

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```

### Example:

````python
messages1 = [
    {"role":"system", "content":"Explain the topic the user wants to understand in a rude/snakry manner while cracking jokes"},
    {"role":"user","content":"Explain about gobal warming in 50 words."}
]
````

````python
response = openai.chat.completions.create(
    model = "gpt-4.1-nano",
    messages=messages1
)
response.choices[0].message.content
````

### Creating a function for taking in system prompt, user prompt and contents of a website for passing it to the model

````python
def message_pack(website):
    return [
        {"role":"system", "content":system_prompt},
        {"role":"user","content":user_prompt + website}
    ]
````

````python
message_pack(content_body)
````

## Mini-Project: Web content summrizer

````python
def summarize(url):
    website  = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = message_pack(website)

    )

    return response.choices[0].message.content

# creating a function to create a clean display for the model output:

def display_text(url):
    summary = summarize(url)
    display(Markdown(summary))
````

````python
display_text("https://edition.cnn.com/")
````

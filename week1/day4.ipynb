# Tokening Text

```python
import tiktoken

text = "Hello, How are you doing today? I hope you're having a great day!"

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

token = encoding.encode(text)

token
```

```python
for token_id in token:
    token_text = encoding.decode([token_id])
    print(f"{token_id}: {token_text}")
```

## Illusion of "Memory"

* LLM models on it's own doesn't have any memory, so if you tried to ask whether it remembers your previous question, it won't rememer at all!

```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv(override=True)

api_key = os.getenv("OPENAI_API_KEY")

if api_key:
    print("Loaded!")
else:
    print("Not Loaded!")
```

```python
from openai import OpenAI

open_ai = OpenAI()
```

```python
messages = [
    {"role":"system", "content":"You are a helpful assistant!"},
    {"role":"user", "content":"What is the capital of France?"}
]
```

```python
response = open_ai.chat.completions.create(
    model = "gpt-5-nano",
    messages = messages
)

response_message = response.choices[0].message.content

display(Markdown(response_message))
```

### Asking the model if it remembers my previous question?

```python
messages = [
    {"role":"system", "content":"You are a helpful assistant!"},
    {"role":"user", "content":"what was the previous question I asked you?"}
]
```

```python
response = open_ai.chat.completions.create(
    model = "gpt-5-nano",
    messages = messages
)

response.choices[0].message.content
```

### Summary:

* Every call to LLM models is stateless (i.e, a clean slate).
* Since every input or prompt provided to LLM models usually contain the entire context.
* That's why it feels as if the LLM models have in-built memory to hold information from past conversations!
